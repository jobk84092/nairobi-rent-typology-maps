# Nairobi Rent Typology: Methodology Walkthrough

This notebook gives a short, portfolio-friendly explanation of how the project estimates and maps rents across Nairobi zones and typologies.

## 1) Load Generated Outputs

The main script writes CSV outputs to the `outputs/` folder. Here we quickly inspect the geo-zone typology table.

```python
from pathlib import Path
import pandas as pd

path = Path('outputs/nairobi_geozone_typology_rent_summary.csv')
df = pd.read_csv(path)
df.head()
```

## 2) Validate Coverage

Quick checks to confirm geographic coverage and typology categories.

```python
summary = {
    'rows': len(df),
    'geo_zones': df['geo_zone'].nunique(),
    'typologies': sorted(df['typology'].unique().tolist()),
}
summary
```

## 3) Compare Typologies

Aggregate citywide average rent by typology to show the expected price ladder.

```python
typology_avg = (
    df.groupby('typology', as_index=False)['avg_rent_ksh']
      .mean()
      .sort_values('avg_rent_ksh')
)
typology_avg
```

## 4) Interpretation Notes

- Values are modeled estimates suitable for analytics demos and portfolio storytelling.
- Boundary geometry comes from the ArcGIS Nairobi layer queried as GeoJSON.
- The script can swap in richer observed rent feeds and finer ward polygons later.
